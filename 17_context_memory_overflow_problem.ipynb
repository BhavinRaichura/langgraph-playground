## Context Memory Overflow

Context memory overflow occurs when the conversation history sent to the LLM becomes too large and exceeds the model's context window.

### Problems
- Increased token usage and cost
- Higher latency
- Context-length errors
- Important information may get buried or ignored

### Solutions
- **Summarization** — Compress older messages.
- **Sliding Window** — Keep only recent messages.
- **Message Trimming** — Remove unnecessary messages.
- **Long-Term Memory** — Store important information separately.
- **RAG** — Retrieve only relevant information when needed.

> **Important:** STM can store a large conversation history, but we should not send the entire history to the LLM on every request.
